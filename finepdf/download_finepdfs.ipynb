```python
from huggingface_hub import snapshot_download

folder = snapshot_download(
    "HuggingFaceFW/finepdfs",
    repo_type="dataset",
    local_dir="./finepdfs/",
    allow_patterns=["data/roh_Latn/train/*"]
)
```

```python
from datasets import load_dataset

DATASET_NAME = "HuggingFaceFW/finepdfs"
CONFIG_NAME = "ita_Latn"
SPLIT_NAME = "train"
FILTER_STRING = "bk.admin.ch"
OUTPUT_FILE = "filtered_urls_ita.txt"

print(f"Loading dataset {DATASET_NAME}...")
# We use streaming=True to avoid downloading the entire dataset if it's large
# The library will intelligently stream and filter.
dataset = load_dataset(
    DATASET_NAME,
    CONFIG_NAME,
    split=SPLIT_NAME,
    streaming=True
)

print(f"Filtering for URLs containing '{FILTER_STRING}'...")
# Define a filter function
def contains_filter(example):
    # Check if 'url' key exists and is not None
    if example.get("url"):
        return FILTER_STRING in example["url"]
    return False

filtered_dataset = dataset.filter(contains_filter)

# Iterate and save the matching URLs
count = 0
print(f"Saving matching URLs to {OUTPUT_FILE}...")
with open(OUTPUT_FILE, 'w', encoding='utf-8') as f:
    for row in filtered_dataset:
        f.write(row["url"] + "\n")
        count += 1

print(f"Done. Found and saved {count} URLs.")
```
