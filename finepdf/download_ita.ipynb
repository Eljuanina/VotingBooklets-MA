```python
from datasets import load_dataset, DownloadConfig

# Configuration
DATASET_NAME = "HuggingFaceFW/finepdfs"
CONFIG_NAME = "ita_Latn"
SPLIT_NAME = "train"
FILTER_STRING = "bk.admin.ch"
OUTPUT_FILE = "filtered_urls_ita.txt"

# Load dataset with automatic cleanup of extracted files
print("Loading dataset")
dataset = load_dataset(
    DATASET_NAME,
    CONFIG_NAME,
    split=SPLIT_NAME,
    download_config=DownloadConfig(delete_extracted=True)
)

print(f"Loaded dataset with {len(dataset)} rows and columns: {dataset.column_names}")

# Confirm that 'url' exists
if "url" not in dataset.column_names:
    raise KeyError(f"Column 'url' not found. Available columns: {dataset.column_names}")

# Filter for rows where 'url' contains the target string (case-insensitive)
print(f"Filtering URLs containing '{FILTER_STRING}'...")
filtered_dataset = dataset.filter(
    lambda row: FILTER_STRING.lower() in row["url"].lower()
)

print(f"Found {len(filtered_dataset)} matching rows.")

# Extract only the 'url' column
urls = filtered_dataset["url"]

# Save URLs to text file, one per line
with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    f.write("\n".join(urls))

print(f"Saved {len(urls)} URLs to '{OUTPUT_FILE}'")
```

```python
from datasets import load_dataset

# Configuration
DATASET_NAME = "HuggingFaceFW/finepdfs"
CONFIG_NAME = "ita_Latn"
SPLIT_NAME = "train"
FILTER_STRING = "www.bk.admin.ch/dam/"
OUTPUT_FILE = "filtered_urls_ita.txt"

print(f"🌐 Streaming dataset {DATASET_NAME}/{CONFIG_NAME} ({SPLIT_NAME})...")
dataset = load_dataset(
    DATASET_NAME,
    CONFIG_NAME,
    split=SPLIT_NAME,
    streaming=True  # no downloading, no caching, constant memory
)

count_total = 0
count_filtered = 0

print(f"Filtering URLs containing '{FILTER_STRING}' and appending to {OUTPUT_FILE}...\n")

for i, example in enumerate(dataset, start=1):
    url = example.get("url")
    pdf_name = example.get("pdf_name") or "unknown_file"

    # progress info
    if i % 1000 == 0:
      print(f"[{i}] {pdf_name}")

    if url and FILTER_STRING.lower() in url.lower():
        with open(OUTPUT_FILE, "a", encoding="utf-8") as f:
            f.write(url + "\n")
        count_filtered += 1
        print(f"   Match #{count_filtered}: {url}")

    count_total += 1

print(f"\n  Done! Processed {count_total} rows, saved {count_filtered} URLs to {OUTPUT_FILE}")
```

```python
from datasets import load_dataset

# Configuration
DATASET_NAME = "HuggingFaceFW/finepdfs"
CONFIG_NAME = "ita_Latn"
SPLIT_NAME = "train"
FILTER_STRING = "www.bk.admin.ch/dam/"
OUTPUT_FILE = "filtered_urls_ita.txt"

print(f"Streaming dataset {DATASET_NAME}/{CONFIG_NAME} ({SPLIT_NAME})...")

# Use streaming mode — reads lazily, doesn't download all
dataset = load_dataset(
    DATASET_NAME,
    CONFIG_NAME,
    split=SPLIT_NAME,
    streaming=True
)

print(f"Filtering URLs containing '{FILTER_STRING}' and writing to {OUTPUT_FILE}...\n")

count_total = 0
count_filtered = 0

# Open file once instead of inside the loop
with open(OUTPUT_FILE, "w", encoding="utf-8") as f_out:
    for i, example in enumerate(dataset, start=1):
        count_total += 1

        url = example.get("url")
        text = example.get("text", "")

        # Progress indicator every 10k lines
        if i % 10_000 == 0:
            print(f"[{i:,}] processed — {count_filtered:,} matches so far...")

        # Filter efficiently, case-insensitive match
        if url and FILTER_STRING.lower() in url.lower():
            f_out.write(f"{url}\t{text.strip()}\n")
            count_filtered += 1

print(f"\n Done! Processed {count_total:,} rows, saved {count_filtered:,} matches to {OUTPUT_FILE}")
```
